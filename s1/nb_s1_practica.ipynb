Ejercicio 1 — Total de pedidos por estado

```python
%sql
SELECT
    status,
    COUNT(*) AS total_orders
FROM training.lesson_01.orders_delta_pair_01
GROUP BY
    status
ORDER BY
    total_orders DESC;
```

Ejercicio 2 — Importe total por país

```python
%sql
SELECT
    country,
    ROUND(SUM(amount), 2) AS total_amount
FROM training.lesson_01.orders_delta_pair_01
GROUP BY
    country
ORDER BY
    total_amount DESC;
```

Ejercicio 3 — País con mayor importe vendido

```python
%sql
SELECT
    country,
    ROUND(SUM(amount), 2) AS total_amount
FROM training.lesson_01.orders_delta_pair_01
GROUP BY
    country
ORDER BY
    total_amount DESC
LIMIT 1;
```

Ejercicio 4 — Ticket medio por canal

```python
%sql
SELECT
    channel,
    ROUND(AVG(amount), 2) AS average_order_amount
FROM training.lesson_01.orders_delta_pair_01
GROUP BY
    channel
ORDER BY
    average_order_amount DESC;
```

Ejercicio 5 — Pedidos cancelados por país

```python
%sql
SELECT
    country,
    COUNT(*) AS cancelled_orders
FROM training.lesson_01.orders_delta_pair_01
WHERE status = 'Cancelled'
GROUP BY
    country
ORDER BY
    cancelled_orders DESC;
```

Ejercicio 6 — Ventas por día

```python
%sql
SELECT
    order_date,
    COUNT(*) AS total_orders,
    ROUND(SUM(amount), 2) AS total_amount
FROM training.lesson_01.orders_delta_pair_01
GROUP BY
    order_date
ORDER BY
    order_date;
```

Ejercicio 7 — Canales con más de 10 pedidos

```python
%sql
SELECT
    channel,
    COUNT(*) AS total_orders,
    ROUND(SUM(amount), 2) AS total_amount
FROM training.lesson_01.orders_delta_pair_01
GROUP BY
    channel
HAVING COUNT(*) > 10
ORDER BY
    total_orders DESC;
```

Ejercicio 8 — Comprobación de nulos

```python
%sql
SELECT
    SUM(CASE WHEN order_id IS NULL THEN 1 ELSE 0 END) AS null_order_id,
    SUM(CASE WHEN customer_id IS NULL THEN 1 ELSE 0 END) AS null_customer_id,
    SUM(CASE WHEN order_date IS NULL THEN 1 ELSE 0 END) AS null_order_date,
    SUM(CASE WHEN country IS NULL THEN 1 ELSE 0 END) AS null_country,
    SUM(CASE WHEN channel IS NULL THEN 1 ELSE 0 END) AS null_channel,
    SUM(CASE WHEN amount IS NULL THEN 1 ELSE 0 END) AS null_amount,
    SUM(CASE WHEN status IS NULL THEN 1 ELSE 0 END) AS null_status
FROM training.lesson_01.orders_delta_pair_01;
```
