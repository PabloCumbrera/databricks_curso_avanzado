## 1. Tipos básicos

Python tiene tipos simples que se usan constantemente en notebooks de datos:

| Tipo | Ejemplo | Uso habitual |
|---|---|---|
| `int` | `10` | contadores, cantidades |
| `float` | `10.5` | importes, métricas |
| `str` | `"ES"` | texto, IDs, países |
| `bool` | `True` | flags, condiciones |
| `None` | `None` | ausencia de valor |

````python
customer_id = "C001"
amount = 120.50
number_of_items = 2
is_active = True
discount_code = None

print(customer_id)
print(amount)
print(number_of_items)
print(is_active)
print(discount_code)

print(type(customer_id))
print(type(amount))
print(type(number_of_items))
print(type(is_active))
print(type(discount_code))
````

## 2. Operadores aritméticos

Operadores principales:

| Operador | Significado |
|---|---|
| `+` | suma |
| `-` | resta |
| `*` | multiplicación |
| `/` | división |
| `//` | división entera |
| `%` | resto |
| `**` | potencia |

````python
price = 100
tax = 21
quantity = 3

total_without_tax = price * quantity
tax_amount = total_without_tax * tax / 100
total_with_tax = total_without_tax + tax_amount

print("Total without tax:", total_without_tax)
print("Tax amount:", tax_amount)
print("Total with tax:", total_with_tax)

print("Integer division:", 10 // 3)
print("Remainder:", 10 % 3)
print("Power:", 2 ** 3)
````

## 3. Operadores de comparación

Devuelven siempre un booleano: `True` o `False`.

````python
amount = 540

print(amount > 500)
print(amount < 500)
print(amount >= 540)
print(amount <= 540)
print(amount == 540)
print(amount != 540)
````

## 4. Operadores lógicos

Permiten combinar condiciones.

````python
amount = 540
status = "Completed"
country = "ES"

is_high_value = amount >= 500
is_completed = status == "Completed"
is_spanish_order = country == "ES"

print(is_high_value)
print(is_completed)
print(is_spanish_order)

print(is_high_value and is_completed)
print(is_high_value or is_spanish_order)
print(not is_completed)
````

## 5. Listas

Una lista guarda varios elementos y mantiene el orden.

````python
amounts = [120.50, 75.00, 60.00, 540.00]

print(amounts)
print(amounts[0])
print(amounts[1])
print(len(amounts))

amounts.append(220.00)

print(amounts)
print(len(amounts))
````

## 6. Tuplas

Una tupla también guarda varios elementos, pero normalmente se usa para valores que no se quieren modificar.

````python
coordinates = (40.4168, -3.7038)

latitude = coordinates[0]
longitude = coordinates[1]

print(coordinates)
print(latitude)
print(longitude)
print(type(coordinates))
````

## 7. Sets

Un set guarda valores únicos. Sirve para validar pertenencia o eliminar duplicados.

````python
countries = {"ES", "PT", "IT", "ES"}

print(countries)
print("ES" in countries)
print("FR" in countries)

valid_statuses = {"Completed", "Pending", "Cancelled"}

status = "Unknown"

print(status in valid_statuses)
````

## 8. Diccionarios

Un diccionario guarda información en formato clave-valor. Es la estructura más parecida a un objeto JSON.

````python
order = {
    "order_id": "O1001",
    "customer_id": "C001",
    "status": "Completed",
    "amount": 120.50,
    "discount_code": None
}

print(order)
print(order["order_id"])
print(order["amount"])

order["classification"] = "Standard"

print(order)
````

## 9. Lista de diccionarios

En datos es muy habitual tener una lista donde cada elemento es un diccionario.

````python
orders = [
    {"order_id": "O1001", "status": "Completed", "amount": 120.50},
    {"order_id": "O1002", "status": "Pending", "amount": 75.00},
    {"order_id": "O1003", "status": "Cancelled", "amount": 60.00},
    {"order_id": "O1004", "status": "Completed", "amount": 540.00}
]

print(orders)
print(orders[0])
print(orders[0]["order_id"])
````

## 10. `if`, `elif`, `else`

Sirven para ejecutar una rama u otra según una condición.

````python
amount = 540

if amount >= 500:
    print("High Value")
else:
    print("Standard")
````

````python
status = "Cancelled"
amount = 60

if status == "Cancelled":
    print("Discarded")
elif amount >= 500:
    print("High Value")
else:
    print("Standard")
````

## 11. Bucle `for`

Sirve para recorrer una colección.

````python
for order in orders:
    print(order["order_id"], order["amount"])
````

````python
for order in orders:
    if order["amount"] >= 500:
        print(order["order_id"], "High Value")
    else:
        print(order["order_id"], "Standard")
````

## 12. Bucle `while`

Se ejecuta mientras una condición sea verdadera.

````python
counter = 1

while counter <= 5:
    print("Counter:", counter)
    counter = counter + 1
````

## 13. `break`

Permite parar un bucle antes de que termine.

````python
for order in orders:
    print("Checking order:", order["order_id"])

    if order["status"] == "Cancelled":
        print("Cancelled order found:", order["order_id"])
        break
````

## 14. Mini-caso combinando varias piezas

Objetivo: recorrer pedidos, descartar cancelados y contar pedidos de alto valor.

````python
valid_orders = []
high_value_orders = []

for order in orders:
    if order["status"] == "Cancelled":
        print(order["order_id"], "is discarded")
        continue

    valid_orders.append(order)

    if order["amount"] >= 500:
        high_value_orders.append(order)

print("Valid orders:", len(valid_orders))
print("High value orders:", len(high_value_orders))
````

## 15. Funciones

Una función permite reutilizar lógica.

````python
def classify_order(order):
    """Classifies an order using status and amount.

    Args:
        order: Order represented as a Python dictionary.

    Returns:
        Order classification.

    Raises:
        KeyError: If status or amount does not exist.
    """
    if order["status"] == "Cancelled":
        return "Discarded"

    if order["amount"] >= 500:
        return "High Value"

    return "Standard"


print(classify_order(orders[0]))
print(classify_order(orders[2]))
print(classify_order(orders[3]))
````

````python
for order in orders:
    classification = classify_order(order)
    print(order["order_id"], classification)
````

## 16. `*args`

`*args` permite recibir varios argumentos posicionales.

````python
def add_amounts(*amounts):
    """Adds any number of amount values.

    Args:
        *amounts: Numeric amount values.

    Returns:
        Sum of all input amounts.

    Raises:
        TypeError: If one amount is not numeric.
    """
    total = 0

    for amount in amounts:
        total = total + amount

    return total


print(add_amounts(10, 20))
print(add_amounts(10, 20, 30, 40))
````

## 17. `**kwargs`

`**kwargs` permite recibir argumentos por nombre.

````python
def build_order_message(**kwargs):
    """Builds a readable order message.

    Args:
        **kwargs: Named order attributes.

    Returns:
        Formatted order message.

    Raises:
        KeyError: If a required named argument is missing.
    """
    order_id = kwargs["order_id"]
    status = kwargs["status"]
    amount = kwargs["amount"]

    return f"Order {order_id} has status {status} and amount {amount}"


message = build_order_message(order_id="O1001", status="Completed", amount=120.50)

print(message)
````

## 18. Lambda

Una lambda es una función pequeña definida en una sola línea.

````python
add_tax = lambda amount: amount * 1.21

print(add_tax(100))
print(add_tax(200))
````

````python
amounts = [120.50, 75.00, 60.00, 540.00]

amounts_with_tax = list(map(lambda amount: amount * 1.21, amounts))

print(amounts)
print(amounts_with_tax)
````

## 19. Scope

El scope define dónde existe una variable.

````python
currency = "EUR"


def print_order_currency():
    """Prints local and global currency examples.

    Args:
        None.

    Returns:
        None.

    Raises:
        None.
    """
    local_message = "This variable exists only inside the function"
    print(currency)
    print(local_message)


print_order_currency()

print(currency)

# This line would fail because local_message only exists inside the function.
# print(local_message)
````

# 20. JSON

A partir de aquí usamos un fichero JSON pequeño.

Fichero esperado en Databricks:

```text
/dbfs/FileStore/tables/m02_python_demo_store.json
```

El fichero tiene:

- `source`
- `load_date`
- `settings`
- `customers`
- `orders`

````python
import json


file_path = "/dbfs/FileStore/tables/m02_python_demo_store.json"

with open(file_path, "r", encoding="utf-8") as file:
    store_data = json.load(file)

print(type(store_data))
print(store_data.keys())
````

## 21. Explorar campos simples del JSON

````python
print(store_data["source"])
print(store_data["load_date"])
print(store_data["settings"])
print(store_data["settings"]["currency"])
print(store_data["settings"]["high_value_threshold"])
````

## 22. Explorar listas dentro del JSON

````python
customers = store_data["customers"]
orders_from_json = store_data["orders"]

print(type(customers))
print(type(orders_from_json))

print(len(customers))
print(len(orders_from_json))

print(customers[0])
print(orders_from_json[0])
````

## 23. Extraer partes del JSON

````python
first_customer = customers[0]
first_order = orders_from_json[0]

print(first_customer["name"])
print(first_customer["country"])
print(first_order["order_id"])
print(first_order["amount"])
print(first_order["items"])
````

## 24. Recorrer pedidos del JSON

````python
for order in orders_from_json:
    print(order["order_id"], order["status"], order["amount"])
````

## 25. Clasificar pedidos del JSON

````python
threshold = store_data["settings"]["high_value_threshold"]

for order in orders_from_json:
    if order["status"] == "Cancelled":
        classification = "Discarded"
    elif order["amount"] >= threshold:
        classification = "High Value"
    else:
        classification = "Standard"

    print(order["order_id"], classification)
````

## 26. Ejercicio guiado dentro de la demo

Completar con el instructor:

1. Contar pedidos completados.
2. Contar pedidos cancelados.
3. Sumar el importe de pedidos completados.

````python
completed_orders = 0
cancelled_orders = 0
completed_amount = 0

for order in orders_from_json:
    if order["status"] == "Completed":
        completed_orders = completed_orders + 1
        completed_amount = completed_amount + order["amount"]

    if order["status"] == "Cancelled":
        cancelled_orders = cancelled_orders + 1

print("Completed orders:", completed_orders)
print("Cancelled orders:", cancelled_orders)
print("Completed amount:", completed_amount)
````

Objetivo: construir una lista con solo los pedidos de alto valor.

````python
high_value_orders = []

for order in orders_from_json:
    if order["amount"] >= store_data["settings"]["high_value_threshold"]:
        high_value_orders.append(order)

for order in high_value_orders:
    print(order["order_id"], order["amount"])
````
